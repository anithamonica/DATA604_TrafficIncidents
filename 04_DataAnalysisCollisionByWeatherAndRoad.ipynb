### Step 4: CollisionByWeatherAndRoad data analysis and visualization

 - How do weather and road alignment influence collision outcomes?



```python
USER = "student"
DB   = "student"
PWD = "Bi3KSjqgrNOOL"

engine = create_engine(
    f"mysql+mysqlconnector://{USER}:{PWD}@127.0.0.1:3306/{DB}"
)

# How do weather and road alignment influence collision outcomes? 
queryMonthly = '''SELECT  Weather, RoadAlignment, SUM(NoofCollisions) AS TotalCollisions,
    ROUND(SUM(NoofInjured) * 1.0 / SUM(NoofCollisions), 2) AS AvgInjuriesPerCollision,
    ROUND(SUM(NoofFatalities) * 1.0 / SUM(NoofCollisions), 4) AS FatalityRate
    FROM student.CollisionByWeatherAndRoad WHERE NoofCollisions > 0 
    AND RoadAlignment <> 'Other' AND Weather <> 'Other'
    GROUP BY Weather, RoadAlignment ORDER BY FatalityRate DESC, AvgInjuriesPerCollision DESC;'''
result = pd.read_sql_query(queryMonthly, engine)
display(result)

pivot = result.pivot(index="RoadAlignment", columns="Weather", values="FatalityRate")
fig, ax = plt.subplots(figsize=(10,6))
cax = ax.matshow(pivot, cmap="viridis")
ax.set_xticks(np.arange(len(pivot.columns)))
ax.set_xticklabels(pivot.columns, rotation=45, ha="left")
ax.set_yticks(np.arange(len(pivot.index)))
ax.set_yticklabels(pivot.index)
plt.colorbar(cax, label="Fatality Rate")
plt.title("Fatality Rate by Weather and Road Alignment")
plt.show()
```

```python

pivot = result.pivot(index="RoadAlignment", columns="Weather", values="AvgInjuriesPerCollision")
fig, ax = plt.subplots(figsize=(10,6))
cax = ax.matshow(pivot, cmap="cividis")
ax.set_xticks(np.arange(len(pivot.columns)))
ax.set_xticklabels(pivot.columns, rotation=45, ha="left")
ax.set_yticks(np.arange(len(pivot.index)))
ax.set_yticklabels(pivot.index)
plt.colorbar(cax, label="AvgInjuriesPerCollision")
plt.title("AvgInjuriesPerCollision  by Weather and Road Alignment")
plt.show()
```

#### Observation From the result / Answer for the query

How do weather and road alignment influence collision outcomes? 

 - Straight and level roads implies largest collision counts, lowest fatality rates.
 - Fatality rate is high if the road alignmet and weather combination is (Visibility limitation, Top of hill or gradient), (Bottom of hill or gradient, Strong wind) or (Clear and sunny, Curved with gradient)
 -  Top or bottom of hills is smaller in collision counts but highest injury severity (up to 1.60 injuries per collision).
 - Clear and sunny weather produces the highest number of collisions, especially on straight and level roads, but fatality rates remain relatively low

### Summary

Most crashes happen on straight, level roads in clear weather, but these are usually the least deadly. Fatality risk rises sharply when poor visibility, strong winds on slopes, or clear weather on curved gradients come into play. Hilltop and downhill sections see fewer crashes overall, yet they cause the most severe injuries. In short, clear weather drives collision numbers, while adverse conditions and tricky road alignments make crashes far more dangerous.
