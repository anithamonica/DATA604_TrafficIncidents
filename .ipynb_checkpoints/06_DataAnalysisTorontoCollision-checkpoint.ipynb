# Comparative Analysis of Toronto Traffic Collisions with NCDB collisions 

As Toronto consistently ranks highest in total reported traffic collisions among Canadian cities, the collision dataset provided by the Toronto police Service Records Management systems licensed by Open Government License - Toronto, is also used for comparative analysis. But the data set contains records from 2014 to current year. Since the NCDB dataset considered for analysis is from 2011 to 2021, I have only included the commom years in both the datasets that is 2014 to 2021.

Here I have considered the below queries for the comparative analysis:

 - Do Toronto’s peak hours/months/day of weeks differ from overall patterns?
 - Compare weather-impacted months (do Toronto’s weather collisions align with national averages?)
 - Are Toronto’s urban road class collisions match national distributions

### Step 3: Data preprocessing

```python
#Introduce new Column : Season
#----------------------------------------------------------------------------
# Mapping from month name to season
monthToSeason = {
    'December': 'Winter', 'January': 'Winter', 'February': 'Winter',
    'March': 'Spring', 'April': 'Spring', 'May': 'Spring',
    'June': 'Summer', 'July': 'Summer', 'August': 'Summer',
    'September': 'Fall', 'October': 'Fall', 'November': 'Fall'
}
dfToronto['Season'] = dfToronto['Month'].map(monthToSeason)
print("Column 'Season' is inserted")
display(dfToronto.head(5))
```

```python
# Insert into table TorontoCollisions
insertCursor = myconnection.cursor()

try:
    myconnection.start_transaction()

    for i, currentRow in dfToronto.iterrows():
        insertCommand = "INSERT INTO `TorontoCollisions` VALUES (" + "%s," * (len(currentRow) - 1) + "%s)"
        insertCursor.execute(insertCommand, tuple(currentRow))

    myconnection.commit()
    print("TorontoCollisions: Inserted records successfully")

except Exception as e:
    myconnection.rollback()
    print("Transaction rolled back due to error:", e)

finally:
    insertCursor.close()
```

### Step 4: TorontoCollisions data analysis and visualization

 - Do Toronto’s peak hours/months/day of weeks differ from overall patterns?
 - Compare weather-impacted months (do Toronto’s weather collisions align with national averages?)




```python
USER = "student"
DB   = "student"
PWD = "Bi3KSjqgrNOOL"

engine = create_engine(
    f"mysql+mysqlconnector://{USER}:{PWD}@127.0.0.1:3306/{DB}"
)
```

```python
#a)	Do Toronto’s peak hours/months/day of weeks differ from overall patterns?
queryMonthly = '''SELECT Month, SUM(CountofCollisions) AS TotalCollisions,
    SUM(NoofFatalities) AS TotalFatalities, SUM(CountofInjuryCollsn) AS TotalInjuryCollisions
    FROM student.TorontoCollisions GROUP BY Month ORDER BY TotalCollisions DESC;'''
monthlyCollisions = pd.read_sql_query(queryMonthly, engine)
display(monthlyCollisions)

queryHour = '''SELECT Timesegment, SUM(CountofCollisions) AS TotalCollisions,
    SUM(NoofFatalities) AS TotalFatalities, SUM(CountofInjuryCollsn) AS TotalInjuryCollisions
    FROM student.TorontoCollisions GROUP BY Timesegment ORDER BY TotalCollisions DESC;'''
hourlyCollisions = pd.read_sql_query(queryHour, engine)
display(hourlyCollisions)

querydayOfWeek = '''SELECT DayofWeek, SUM(CountofCollisions) AS TotalCollisions,
    SUM(NoofFatalities) AS TotalFatalities, SUM(CountofInjuryCollsn) AS TotalInjuryCollisions
    FROM student.TorontoCollisions GROUP BY DayofWeek ORDER BY TotalCollisions DESC;'''
dayCollisions = pd.read_sql_query(querydayOfWeek, engine)
display(dayCollisions)
```

#### Observation From the result / Answer for the query

Do Toronto’s peak hours/months/day of weeks differ from overall patterns?

- The peak collision months for Toronto are October to January and July seems moderately high. Also the collision count dips in March to May. 
- Late afternoon has more collisions.
- Fridays and Thursdays are having more collisions

The NCDB collision analysis has the following results:

Collisions peak in July, August, November months and dip in February, March, April months. Weekdays, especially Thursdays and Fridays, see the highest collision counts, with most accidents occurring between mid‑morning and evening, particularly during the late afternoon rush (3–6 PM)

From this we can conclude that the Toronto’s peak hours/months/day of weeks align with National collision data.


```python
#b)	Compare weather-impacted months (do Toronto’s weather collisions align with national averages?)
seasonMap = { "December": "Winter", "January": "Winter", "February": "Winter",
              "March": "Spring", "April": "Spring", "May": "Spring", "June": "Summer",
              "July": "Summer", "August": "Summer",  "September": "Fall", "October": "Fall", "November": "Fall"}
monthlyCollisions["Season"] = monthlyCollisions["Month"].map(seasonMap)

seasonSummary = monthlyCollisions.groupby("Season").agg({"TotalCollisions": "sum", "TotalFatalities": "sum", 
                                                          "TotalInjuryCollisions": "sum"}).reset_index()
season_avg = (monthlyCollisions.groupby("Season")[["TotalCollisions","TotalFatalities","TotalInjuryCollisions"]]
    .mean().reset_index().sort_values(by="TotalCollisions", ascending=False))

display(season_avg)
```

#### Observation From the result / Answer for the query

Compare weather-impacted months (do Toronto’s weather collisions align with national averages?)

 For national Collisions peak in Fall and Summer, dip in Spring, while injury severity is steady and fatalities remain low but stable across seasons. Fall and summer are the riskiest seasons with highest collision counts. Winter has fewer fatalities relative to collisions, suggesting safer outcomes despite challenging conditions

 From the above table we can see that average Toronto collisions peak in Fall and Winter months. But still the average fatality is more on fall and summer as well as the average injuries. So we can say Toronto’s seasonal collision averages align closely with national patterns.
